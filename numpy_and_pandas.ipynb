```python

df=pd.read_csv('F:\Data Science\car-sales-extended-missing-data.csv')
df.head()
```

```python
make_grouped=df.groupby('Make')['Doors'].mean()
make_grouped
```

```python
sum_grouped=df.groupby(['Colour','Make'])['Doors'].sum()
sum_grouped
```

```python
df1=pd.DataFrame({
    'Key':['a','b','c'],
    'Value':[1,2,3]
})
df2=pd.DataFrame({
    'Key':['a','b','d'],
    'Value':[1,2,4]
})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on="Key",how="inner")
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```
