### importing sqlite

```python
import sqlite3
```

### Connecting to sqlite

```python
connection=sqlite3.connect("sample.db")
connection
```

### Creating cursor


```python
cursor=connection.cursor()
cursor
```

### Creating a table

```python
cursor.execute('''
    Create Table If Not Exists Employees(
        id Integer Primary Key,
        name Text not Null,
        age Integer,
        department Text
    )
''')

##Commiting the changing(Saving permanently)
connection.commit()
```

### Inserting values in table


```python
cursor.execute('''
    Insert Into Employees(id,name,age,department)
    values(1,"Bhoomi",22,"cse")
    
''')
cursor.execute('''
    Insert Into Employees(id,name,age,department)
    values(2,"Ajay",23,"cse")
    
''')
cursor.execute('''
    Insert Into Employees(id,name,age,department)
    values(3,"Tejas",21,"ece")
    
''')


connection.commit()
```

### Fetching values from table

```python
cursor.execute('''
    Select *from Employees
''')
rows=cursor.fetchall()

for row in rows:
    print(row)


connection.commit()
```

### Updating the data

```python
cursor.execute('''
UPDATE Employees
Set name='Bhoomika Somavarapu'
Where name='Bhoomi'
''')

connection.commit()
```

```python
cursor.execute('''
    Select *from Employees
''')
rows=cursor.fetchall()

for row in rows:
    print(row)


connection.commit()
```

### Deleting data from db

```python
cursor.execute('''
    Delete From Employees
    Where age=22
''')

connection.commit()
```

```python
cursor.execute('''
    Select *from Employees
''')
rows=cursor.fetchall()

for row in rows:
    print(row)


connection.commit()
```

# Performing operations in Sales data

```python
cursor.execute('''
    Create Table If Not Exists Sales(
        id Integer Primary Key,
        product Text,
        date Text,
        cost Integer
    )
''')

sales_data=[
    (1,"Keyboard",'10-02-2025',1000),
    (2,"Mouse",'11-02-2025',1000),
    (3,"Monitor",'12-02-2025',1000),
]

connection.commit()
```

```python
cursor.executemany('''
Insert Into Sales(id,product,date,cost)
            values(?,?,?,?)
''',sales_data)
```

```python
cursor.execute('''
    Select *from Sales
''')
rows=cursor.fetchall()

for row in rows:
    print(row)


connection.commit()
```
