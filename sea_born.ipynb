```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=sns.load_dataset('iris')
data
```

```python
##Scatterplot

sns.scatterplot(x='sepal_length',y='sepal_width',data=data)
```

```python
##Line plot

sns.lineplot(x='sepal_length',y='sepal_width',data=data)
```

```python
#Bar Plot
sns.barplot(x='species',y='petal_length',data=data)
```

```python
#box plot
sns.boxplot(x='species',y='petal_length',data=data)
```

```python
#Voilen  plot
sns.violinplot(x='species',y='petal_length',data=data)
```

```python
#Histogram
sns.histplot(data['petal_length'],bins=10,kde=True)
```

```python
#KDE plot

sns.kdeplot(data['petal_length'],shade=True)
```

```python
#Pair plot

sns.pairplot(data)
```

```python
#Heatmap
correl=data[['petal_length','petal_width','sepal_length']].corr()
correl
```
