```python
import matplotlib as mlt
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y)
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]

plt.plot(x,y)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title("Practice plot")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]

plt.plot(x,y,marker='o',linestyle='--',linewidth='4',markersize=10,color='red')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title("Practice plot")
plt.grid(True)
```

```python
#Multiple plots

x=[1,2,3,4,5]
y1=[1,2,9,16,25]
y2=[1,2,3,4,5]

#plot 1
plt.subplot(2,2,1)
plt.plot(x,y1,color='red')

#plot 2
plt.subplot(2,2,2)
plt.plot(x,y2,color='red')

#plot 3
plt.subplot(2,2,3)
plt.plot(x,y1,color='blue')

#plot 4
plt.subplot(2,2,4)
plt.plot(x,y2,color='green')
```

```python
#Bar graph

students=['Bhoomika','Geethike','Ajay','Tejas']
marks=[90,80,70,85]

plt.bar(students,marks,color='orange')
```

```python
#Histogram



age=[1,2,3,4,5,6,7,8,9]
plt.hist(age,bins=5,edgecolor='black')
```

```python
#Scatter plot
x=[1,2,3,4,5]
y=[1,4,9,16,25]

plt.scatter(x,y,marker='x')
```

```python
#Pie chart

labels=['Study','Eating','Exercise','sleep','playing']
sizes=[30,10,10,20,30]
explode=[0.2,0,0,0,0]
colours=['red','green','blue','orange','gold']

plt.pie(sizes,labels=labels,explode=explode,autopct="%1.1f%%",shadow=True)
```
