```python
import logging
```

```python
logging.basicConfig(level=logging.DEBUG)
```

```python
logging.debug("This is Debug log")
logging.info("This is info log")
logging.warning("This is warning log")
logging.error("This is error log")
logging.critical("This is critical log")
```

```python
logging.basicConfig(
    filename='loggingFile.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%D-%M-%Y %H:%M:%S',
    force=True
)

logging.debug("This is Debug log")
logging.info("This is info log")
logging.warning("This is warning log")
logging.error("This is error log")
logging.critical("This is critical log")
```

# Logging with multiple loggers

```python
loggerm1=logging.getLogger('logger1')
loggerm2=logging.getLogger('logger2')

logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%D-%M-%Y %H:%M:%S',
    force=True
)

loggerm1.debug("This is Debug log")
loggerm2.info("This is info log")
```
