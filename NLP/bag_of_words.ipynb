```python
from sklearn.feature_extraction.text import CountVectorizer

documents=[
    "I love dosa",
    "I eat dosa",
    "I love paratha",
    "I eat paratha"
]

vectorizer=CountVectorizer()

res=vectorizer.fit_transform(documents)


print(vectorizer.get_feature_names_out())
print(res.toarray())
```

## using n-grams

```python
from sklearn.feature_extraction.text import CountVectorizer

documents=[
    "I love dosa",
    "I eat dosa",
    "I love paratha",
    "I eat paratha"
]

vectorizer=CountVectorizer(ngram_range=(2,2))

res=vectorizer.fit_transform(documents)


print(vectorizer.get_feature_names_out())
print(res.toarray())
```
