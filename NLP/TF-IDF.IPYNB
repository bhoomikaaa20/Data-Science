```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents=[
    "I love dosa",
    "I eat dosa",
    "I love paratha",
    "I eat paratha"
]

vectorizer=TfidfVectorizer()

res=vectorizer.fit_transform(documents)


print(vectorizer.get_feature_names_out())
print(res.toarray())
```
